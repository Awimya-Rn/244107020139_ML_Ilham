```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.metrics import confusion_matrix, accuracy_score
```

# Pemilihan Fitur

```python
df_train=pd.read_csv('data/fashion-mnist_train.csv')
df_test=pd.read_csv('data/fashion-mnist_test.csv')

X_train = df_train.iloc[:, 1:].values
y_train = df_train.iloc[:, 0].values
X_test = df_test.iloc[:, 1:].values
y_test = df_test.iloc[:, 0].values

print(X_train.shape, X_test.shape)
```

# Metode pra pengolahan

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

# Arsitektur ANN & Fungsi Aktivasi
- activation='relu': Digunakan pada 6 hidden layer pertama. Berfungsi menghitung f(x)=max(0,x) untuk mencegah vanishing gradient dan mempercepat komputasi.
- activation='tanh': Digunakan pada 2 hidden layer berikutnya. Menghasilkan output di rentang [−1,1] (zero-centered).
- activation='sigmoid': Digunakan pada 2 hidden layer lanjutan. Menghasilkan output nonlinear rentang (0,1).
- activation='softmax': Digunakan pada layer output untuk mengubah logit menjadi distribusi probabilitas 10 kelas (total probabilitas = 1).

  

  

```python
ann = tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=64, activation='relu'),
    tf.keras.layers.Dense(units=128, activation='relu'),
    tf.keras.layers.Dense(units=256, activation='relu'),
    tf.keras.layers.Dense(units=128, activation='tanh'),
    tf.keras.layers.Dense(units=64, activation='tanh'),
    tf.keras.layers.Dense(units=128, activation='sigmoid'),
    tf.keras.layers.Dense(units=64, activation='sigmoid'),
    tf.keras.layers.Dense(units=10, activation='softmax')
])
```

```python
ann.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
ann.fit(X_train, y_train, batch_size=32, epochs=20, validation_split=0.1, verbose=0)
```

```python
y_pred = ann.predict(X_test)
y_pred = np.argmax(y_pred, axis=1)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
print('Akurasi:', accuracy_score(y_test, y_pred))
```
