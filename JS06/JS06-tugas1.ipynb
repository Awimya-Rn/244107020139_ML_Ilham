```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

df = pd.read_csv('data/voice.csv')
X = df.drop(columns='label')
y = df['label']
```

```python
kernels = {
  'linear': SVC(kernel='linear', C=1.0),
  'poly': SVC(kernel='poly', degree=3, C=1.0),
  'rbf': SVC(kernel='rbf', C=1.0, gamma='scale')
}
splits_ratio = {
  '70:30':0.3,
  '80:20':0.2
}
```

```python
rows = []
for splits_name, tests_size in splits_ratio.items():
  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=tests_size, random_state=42, stratify=y)
  for kernel_name, model in kernels.items():
    pipeline = make_pipeline(StandardScaler(), model)
    pipeline.fit(X_train, y_train)
    y_pred_train = pipeline.predict(X_train)
    y_pred_test = pipeline.predict(X_test)
    rows.append({
      'split': splits_name,
      'kernel': kernel_name,
      'akurasi_train': accuracy_score(y_train, y_pred_train),
      'akurasi_test': accuracy_score(y_test, y_pred_test)
    })
```

```python
result = pd.DataFrame(rows)
print(result.round(4).to_string(index=False))
print()
print((result.pivot(index='kernel', columns='split', values='akurasi_test')*100).round(2))
```
